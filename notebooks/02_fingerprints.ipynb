Huellas por pares de picos espectrales. Se aprende el anuncio A en el episodio 1 y se reconoce en el 2 **sin transcribir**.

```python
import sys; sys.path.insert(0, '.')
from _common import *
import numpy as np
from podcleany import audio, fingerprint as fp
W = workdir()
def load(src, name):
    audio.normalize_to_blocks(src, W/name, 16000, 60)
    return np.concatenate([audio.read_block(W/name, i) for i in range(audio.count_blocks(W/name))]).astype('float32')/32768
e1, e2 = load(W/'web'/'ep1.mp3','b1'), load(W/'web'/'ep2.mp3','b2')
ad = e1[int(39.5*16000):int(59.5*16000)]
h = fp.hashes_from_audio(ad); print('hashes del anuncio:', len(h))
idx = fp.build_index([(1,'ad',20.0,fp.pack(h))])
print(fp.match(fp.hashes_from_audio(e2), idx))
```
